## 2. Make a bank account

Using what you've learned about object oriented programming, make a class that works as a bank account.

Your banking account should have the following properties (and any other functionality you can think of):
- When opened (initialized) it should have a starting balance of 0 
- The owner should be able to 
    - deposit money (add money to the balance)
    - withdraw money, as long as the withdraw amount doesn't put them over their overdraft limit
    - increase their overdraft limit (starts at 0) by 100 each time

```python
class BankAccount:
    """A simple bank account with deposit, withdraw, and overdraft features."""

    def __init__(self, balance: float = 0) -> None:
        """Initialize a new account with an optional starting balance."""
        self.balance = balance
        self.overdraft_limit = 0

    def __repr__(self) -> str:
        """Return a developer-friendly representation."""
        return (
            f"BankAccount(balance={self.balance}, "
            f"overdraft_limit={self.overdraft_limit})"
        )

    def deposit(self, amount: float) -> None:
        """Add money to the account balance."""
        if amount <= 0:
            raise ValueError(f"Deposit amount must be positive, got {amount}.")
        self.balance += amount

    def withdraw(self, amount: float) -> None:
        """Take money from the account, respecting the overdraft limit."""
        if amount <= 0:
            raise ValueError(f"Withdraw amount must be positive, got {amount}.")
        if amount > self.balance + self.overdraft_limit:
            raise ValueError(
                f"Cannot withdraw {amount}: only {self.balance + self.overdraft_limit} "
                f"available (balance {self.balance} + overdraft {self.overdraft_limit})."
            )
        self.balance -= amount

    def increase_overdraft(self) -> None:
        """Increase the overdraft limit by 100."""
        self.overdraft_limit += 100
```

```python
account = BankAccount()
print(account)                    # BankAccount(balance=0, overdraft_limit=0)

account.deposit(500)
print(account)                    # BankAccount(balance=500, overdraft_limit=0)

account.withdraw(200)
print(account)                    # BankAccount(balance=300, overdraft_limit=0)

# Try to overdraw — should raise
try:
    account.withdraw(400)
except ValueError as e:
    print(f"❌ {e}")                # Cannot withdraw 400: only 300 available...

# Increase overdraft and try again
account.increase_overdraft()       # limit now 100
account.withdraw(400)              # now allowed: balance 300 + overdraft 100 = 400
print(account)                    # BankAccount(balance=-100, overdraft_limit=100)

# Edge case: try negative deposit
try:
    account.deposit(-50)
except ValueError as e:
    print(f"❌ {e}")                # Deposit amount must be positive, got -50.
```
