## Parent Classes and Dunder Methods

Remember the parent class `Deck`? And its possible children? This notebook is here for you to implement some advanced dunder methods.

```python
import collections

Card = collections.namedtuple('Card', ['rank', 'suit'])


class Deck:

    def __init__(self):
        self.cards = [
            Card(rank, suit)
            for suit in self.suits
            for rank in self.ranks
        ]
        self.dealt_cards = []

    def __len__(self):
        return len(self.cards)
        
    def __str__(self):
        return f'Deck(suits={self.suits}, ranks={self.ranks})'
    
    def __getitem__(self, position):
        return self.cards[position]
    
    def __setitem__(self, ind, val):
        self.cards[ind] = val

    def __add__(self, other: "Deck") -> "Deck":
        """Combine two decks into a new deck containing all their cards."""
        if not isinstance(other, Deck):
            return NotImplemented
        new_deck = Deck.__new__(type(self))   # create empty instance of same type
        new_deck.cards = self.cards + other.cards
        new_deck.dealt_cards = []
        return new_deck

    def deal(self):
        return self.cards.pop()

class AvatarDeck(Deck):
    ranks = list('23456789T') + ['Sokka', 'Katara', 'Zuko', 'Aang']
    suits = ['Water', 'Air', 'Fire', 'Water']
            
class French52Deck(Deck):
    ranks = '23456789TJQKA'
    suits = '♠♥♦♣'
    
    def top_card_is_ace(self):
        return self.cards[-1].rank == 'A'

avatar_deck = AvatarDeck()
deck = French52Deck()
```

2. Make a class called `Dealer` that takes a Deck, shuffles it and is prepared to deal `n` number of cards. (For example in a game of 7-card rummy, the players each get 7 cards once the deck is shuffled).

```python
import random


class Dealer:
    """A dealer that manages a deck and deals cards to players."""

    def __init__(self, deck: Deck) -> None:
        """Initialize the dealer with a deck and shuffle it."""
        self.deck = deck
        random.shuffle(self.deck)

    def __repr__(self) -> str:
        """Return a developer-friendly representation."""
        return f"Dealer({len(self.deck)} cards remaining)"

    def deal_hand(self, n: int) -> list[Card]:
        """Deal n cards from the deck and return them as a hand."""
        if n > len(self.deck):
            raise ValueError(
                f"Cannot deal {n} cards — only {len(self.deck)} left."
            )
        return [self.deck.deal() for _ in range(n)]

    def deal_hands(self, n_players: int, cards_per_hand: int) -> list[list[Card]]:
        """Deal hands to multiple players."""
        total_needed = n_players * cards_per_hand
        if total_needed > len(self.deck):
            raise ValueError(
                f"Cannot deal {n_players} hands of {cards_per_hand} — "
                f"need {total_needed} cards, only {len(self.deck)} left."
            )
        return [self.deal_hand(cards_per_hand) for _ in range(n_players)]
```

```python
deck = French52Deck()
dealer = Dealer(deck)
print(dealer)                           # Dealer(52 cards remaining)

# Deal one hand of 7 cards (rummy)
my_hand = dealer.deal_hand(7)
print(my_hand)
print(dealer)                           # Dealer(45 cards remaining)

# Deal hands to 4 players
deck2 = French52Deck()
dealer2 = Dealer(deck2)
hands = dealer2.deal_hands(n_players=4, cards_per_hand=7)
for i, hand in enumerate(hands, start=1):
    print(f"Player {i}: {hand}")
print(dealer2)                          # Dealer(24 cards remaining)
```
